```python
import csv 

def read_csv_manual(filepath):
    result = []
    with open (filepath, 'r', encoding = 'utf-8') as file:
        csv_reader = csv.DictReader(file)
        for row in csv_reader:
            result.append(row)
    return result

result = read_csv_manual('supermarket_sales.csv')
for i in range(2): 
    print(result[i])
```

```python
rewiews = ['great product', 'bad quality great', 'average product', 'great great value']
word_count = {}
for rewiew in rewiews:
    words = rewiew.lower().split()
    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
print(word_count)
```

```python
import pandas as pd 

file = pd.read_csv('supermarket_sales.csv')
price = [price for price in file['Unit price'] if price > file['Unit price'].mean()]
print(file['Unit price'].mean())
print(price)
```

```python
import csv 

result = {'A': 0, 'B': 0, 'C': 0}
with open ('supermarket_sales.csv', 'r', encoding = 'utf-8') as file:
    csv_reader = csv.DictReader(file)
    for row in csv_reader:
        branch = row['Branch']
        total = float(row['Total'])
        result[branch] += total
print(result)    
```
